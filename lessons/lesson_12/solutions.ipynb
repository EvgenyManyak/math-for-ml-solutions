# Практические задания

#### 1. Свой attention на NumPy. Реализуйте attention(Q, K, V) за 5 строк. Прогоните на случайных Q, K, V и сравните с PyTorch.

#### 2. Multi-head на NumPy. Расширьте до multi-head. Покажите, что h=1 совпадает с обычным attention.

#### 3. Position encoding визуализация. Постройте heatmap синусоидального PE для длины 100 и размерности 64. Должны увидеть полосатый красивый узор.

#### 4. Mini-transformer. Реализуйте один блок декодера на PyTorch (без nn.Transformer). Обучите на задаче «сложить два числа» в формате токенов: 12 + 34 = 46.

#### 5. Маска самовнимания. Реализуйте causal mask для декодера. Покажите, что без маски модель «жульничает» (видит будущее) и идеально предсказывает train, но плохо test.

#### 6. Scaling 1/√d важен. Запустите attention с d=64 без масштабирования. Покажите, что softmax становится почти one-hot. Включите масштабирование — должно стать гладко.

#### 7. Своё RoPE. Реализуйте rotary positional embedding и сравните с sinusoidal на коротких последовательностях. Какое быстрее обобщается на длинные?

#### 8. nanoGPT-clone. Возьмите karpathy/nanoGPT как референс. Реализуйте свою версию на 200 строк PyTorch и обучите на Шекспире. Должны получить связный текст.
